!!! info 1 BOUCLES BORNEES : FOR

<div class="alert alert-warning" style="border-left:15px solid #eba83a;">
      
La boucle **_for_** prend ses valeurs dans un **itérable**, par exemple renvoyé par la fonction range() :

```python
for variable in range(10):
    ...
```    
    
À chaque itération la variable prendra pour valeur un élément renvoyé par range()

```python
def table(n = 10):
    texte = ""
    for i in range(1, 11):
        texte += f"{i}*{n}={i*n}\n"
    return texte

print(table())
print(table(6))
